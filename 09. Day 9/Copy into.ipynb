```python
%sql
USE CATALOG wns24082026;

DROP TABLE IF EXISTS quickstart_schema.users;

CREATE TABLE If not EXISTS quickstart_schema.users (
    id INT,
    name STRING,
    dob DATE,
    email STRING,
    gender STRING,
    country STRING,
    region STRING,
    city STRING,
    asset INT,
    marital_status STRING
);
```

# Copy into using manual schema Evolution

## Data Ingestion using Copy Into

```python
%sql
COPY INTO quickstart_schema.users FROM (
  SELECT
    id::INT,
    name::STRING,
    to_date(dob, 'yyyy-MM-dd') as dob,
    email::STRING,
    gender::STRING,
    country::STRING,
    region::STRING,
    city::STRING,
    asset::INT,
    marital_status::STRING FROM
  '/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/input/'
) FILEFORMAT = CSV FORMAT_OPTIONS ('header' = 'true')
```

## Data Ingestion

```python
%sql
COPY INTO quickstart_schema.users FROM
'/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/Input2/' FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true', 'timestampFormat' = 'yyyy-MM-dd', 'inferSchema' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
```
