# Prerequisite

```python
df = spark.read.csv(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/employee.csv",
                    header=True,
                    inferSchema=True,
                    sep="|",
                    quote="'")
display(df)
 
```

# Select Operations

```python
df.select("id","name").show()
```

```python
from pyspark.sql.functions import col
df.select(col("id").alias("user_id"), col("name")).show()
```

# Filter Operations

```python
df.filter(col("desig")=="Developer").show()
```

```python
df.filter((col("desig")=="Developer") | (col("desig")=="Team Lead")).show()
```

```python
df.filter(col("desig").isin("Developer", "Team Lead")).show()
```

# Derive Columns

```python
from pyspark.sql.functions import *
df.withColumn("other", lit("OTH")).show()
```

```python
from pyspark.sql.functions import when
 
result_df = df.withColumn(
    "category",
    when(col("exp") >= 10, "Senior")
    .when(col("exp") >= 5, "Mid Level")
    .when(col("exp") > 0, "Junior")
    .otherwise("Invalid Experience"),
)

result_df.select("exp","category").show()
 
```

```python
result_df.groupBy("category").count().sort(col("count"), ascending=False).show()
```

```python
result_df.groupBy("category").count().sort(col("count").desc()).show()
```

```python
from pyspark.sql.functions import min, max, avg

result_df.groupBy("category").agg(
    min("exp").alias("min_exp"),
    max("exp").alias("max_exp"),
    avg("exp").alias("avg_exp"),
).show()
```
