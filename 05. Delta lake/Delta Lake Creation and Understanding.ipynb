1. Read Delta as DF
2. Create a view on DF
3. Updates are allowed

```python
df = spark.read.parquet(
    "/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet"
)
```

## save delta

```python
df.write.format("delta").save(
    "/Volumes/wns24082026/quickstart_schema/sandbox/output_delta"
)
```

## Read Delta

```python
df_delta = spark.read.format("delta").load(
    "/Volumes/wns24082026/quickstart_schema/sandbox/output_delta")
display(df_delta)
```

## create view

```python
df_delta.createOrReplaceTempView("users_delta_vw")
```

## Update table/view

```python
%sql
update
  users_delta_vw
SET
  country = 'Bharat'
where
  country = 'India'
```

## check view data

```python
%sql
select * from users_delta_vw
```
